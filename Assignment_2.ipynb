## Question 1 — Lists

Take your roll number. Extract its individual digits, and multiply each digit by **10** to form your list `L` (10 numbers in the list now).

**Roll number:** `1024170347`  
**Digits:** `1, 0, 2, 4, 1, 7, 0, 3, 4, 7`  
**L (digit × 10):** `[10, 0, 20, 40, 10, 70, 0, 30, 40, 70]`

### 1(i) Print L

```python
roll_no = "1024170347"
digits = [int(ch) for ch in roll_no]          # extract individual digits
L = [d * 10 for d in digits]                  # multiply each digit by 10

print("Roll number :", roll_no)
print("Digits      :", digits)
print("L           :", L)
print("Length of L :", len(L))
```

### 1(ii) Add two numbers — `append()` and `insert()`

- `append(x)` adds `x` at the **end** of the list.
- `insert(i, x)` adds `x` at **index `i`**, shifting later elements one place to the right.

```python
print("L before adding :", L)

L.append(25)                                  # 25 is added at the end (new last index)
print("After append(25) :", L)                # length grew by 1; only the tail changed

L.insert(3, 15)                               # 15 is inserted at index 3
print("After insert(3, 15) :", L)             # 15 sits at index 3; original index-3 and after shift right
```

### 1(iii) Remove two elements — `remove()` and `pop()`

- `remove(value)` deletes the **first occurrence** of that value.
- `pop()` (no argument) deletes and **returns** the last element.

```python
print("L before removing :", L)

L.remove(0)                                   # removes the first 0 found, not every 0
print("After remove(0) :", L)

popped = L.pop()                              # pops the last element and returns it
print("After pop()     :", L)
print("Value popped    :", popped)
```

### 1(iv) Sort ascending, then descending

```python
L.sort()                                      # in-place ascending sort
print("Ascending  (sort())              :", L)

L.sort(reverse=True)                          # in-place descending sort
print("Descending (sort(reverse=True))  :", L)
```

### 1(v) Slicing — first three and last three

```python
print("First three elements :", L[:3])        # indices 0, 1, 2
print("Last three elements  :", L[-3:])       # last three, in current order
```

### 1(vi) List comprehension — elements greater than the average of L

```python
avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]

print("L              :", L)
print("Average of L   :", avg)
print("Elements > avg :", above_avg)
```

## Question 2 — Tuples

Create a tuple named `scores` containing **8 marks**: use the **first 8 values** from the original list `L` in Q1 (before any later modifications).

Original L from Q1(i): `[10, 0, 20, 40, 10, 70, 0, 30, 40, 70]`  
So `scores = (10, 0, 20, 40, 10, 70, 0, 30)`

```python
# Rebuild the original L from the roll number so Q2 does not depend on later Q1 edits
original_L = [int(ch) * 10 for ch in "1024170347"]
scores = tuple(original_L[:8])                # first 8 values of original L

print("Original L :", original_L)
print("scores     :", scores)
print("type       :", type(scores))
```

### 2(i) Highest score + index; lowest score + frequency

```python
highest = max(scores)
highest_index = scores.index(highest)         # first occurrence of the maximum

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score :", highest, " at index", highest_index)
print("Lowest score  :", lowest,  " appears", lowest_count, "time(s)")
```

### 2(ii) Reverse the tuple and return the result as a list

```python
# Tuples are immutable, so they have no in-place reverse(); we build a new reversed sequence instead.
reversed_as_list = list(reversed(scores))     # or: list(scores[::-1])

print("Original tuple :", scores)
print("Reversed list  :", reversed_as_list)
```

### 2(iii) Search for a user-entered score

```python
user_score = int(input("Enter a score to search in the tuple: "))

if user_score in scores:
    print("First occurrence of", user_score, "is at index", scores.index(user_score))
else:
    print(user_score, "is not present in the tuple scores.")
```

### 2(iv) Attempt to change one element of the tuple

```python
print("scores before attempted change :", scores)

try:
    scores[0] = 100                           # this must fail
except TypeError as err:
    print("Error raised :", type(err).__name__ + ":", err)

# Item assignment fails because a tuple is immutable; a list would accept scores[0] = 100 in place.
print("scores after attempted change  :", scores)
```

### 2(v) Unpack with the `*` operator

```python
first, second, *remaining = scores            # first score, second score, and the rest

print("First score     :", first)
print("Second score    :", second)
print("Remaining scores:", remaining)
```

## Question 3 — Random numbers (seeded by roll number)

Set the random seed to your own roll number using `random.seed(your_roll_number)` **before** generating numbers. This keeps the list random-looking but **reproducible and unique to you**.

### 3(i) Generate 100 random numbers between 100 and 900 (inclusive)

```python
import random
from collections import Counter

random.seed(1024170347)                       # unique, reproducible seed = my roll number

rand_list = [random.randint(100, 900) for _ in range(100)]

print("Seed used          : 1024170347")
print("Length of list     :", len(rand_list))
print("Min / Max in list  :", min(rand_list), "/", max(rand_list))
print("Generated list     :")
print(rand_list)
```

### 3(ii) Count and print all odd numbers

```python
odd_numbers = [n for n in rand_list if n % 2 != 0]

print("Odd numbers :")
print(odd_numbers)
print("Count of odd numbers :", len(odd_numbers))
```

### 3(iii) Count and print all even numbers

```python
even_numbers = [n for n in rand_list if n % 2 == 0]

print("Even numbers :")
print(even_numbers)
print("Count of even numbers :", len(even_numbers))
```

### 3(iv) Count and print all prime numbers (list built with a comprehension)

```python
def is_prime(n):
    """Return True if n is a prime number."""
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2
    return True

prime_numbers = [n for n in rand_list if is_prime(n)]   # list of primes via comprehension

print("Prime numbers :")
print(prime_numbers)
print("Count of prime numbers :", len(prime_numbers))
```

### 3(v) Most frequent number and its frequency

```python
freq = Counter(rand_list)
most_common_num, most_common_count = freq.most_common(1)[0]

print("Number that occurs most frequently :", most_common_num)
print("How many times it occurs           :", most_common_count)
```

## Question 4 — Sets

Using the **same 8 digits** of the roll number from Q1 (before multiplying by 10), create:

- set **A** = `{ digit × 7  for each digit }`
- set **B** = `{ digit × 9  for each digit }`

First 8 digits of `1024170347`: `1, 0, 2, 4, 1, 7, 0, 3`

```python
first_eight_digits = [int(ch) for ch in "1024170347"[:8]]

A = {d * 7 for d in first_eight_digits}
B = {d * 9 for d in first_eight_digits}

print("First 8 digits :", first_eight_digits)
print("Set A (digit × 7) :", A)
print("Set B (digit × 9) :", B)
```

### 4(vi) Union of A and B

```python
union_AB = A.union(B)                         # or: A | B
print("A ∪ B :", union_AB)
print("All unique values across both sets; duplicates appear only once.")
```

### 4(vii) Intersection of A and B

```python
inter_AB = A.intersection(B)                  # or: A & B
print("A ∩ B :", inter_AB)
print("Values that appear in both A and B.")
```

### 4(viii) A − B and B − A separately

```python
A_minus_B = A.difference(B)                   # or: A - B
B_minus_A = B.difference(A)                   # or: B - A

print("A − B :", A_minus_B)
print("B − A :", B_minus_A)
# difference() is one-directional (only in the first set); symmetric_difference() is both leftover sides together.
```

### 4(ix) Symmetric difference of A and B

```python
sym_diff = A.symmetric_difference(B)          # or: A ^ B
print("A △ B :", sym_diff)
print("Values in A or B but not in both  ==  (A − B) ∪ (B − A).")
```

### 4(x) Subset / superset checks

```python
print("A issubset  of B :", A.issubset(B))
print("B issuperset of A :", B.issuperset(A))
```

### 4(xi) Remove a user-entered value X from A using `discard()`

```python
X = int(input("Enter a value X to discard from set A: "))

print("A before discard :", A)
A.discard(X)                                  # does nothing if X is not in A
print("A after  discard :", A)
# discard() is safer than remove() because missing values do not raise KeyError.
```

## Question 5 — Dictionaries

Consider the following dictionary built from the personal-details cell at the top of this notebook.

```python
my_dict = {
    "name": STUDENT_NAME,
    "roll_no": STUDENT_ROLL,
    "branch": STUDENT_BRANCH,
    "age": STUDENT_AGE,
    "city": STUDENT_CITY,
}

print("Original dictionary:")
for k, v in my_dict.items():
    print(f"  {k!r}: {v!r}")
```

### 5(i) Rename the key `"city"` to `"location"` without recreating the dictionary by hand

```python
# pop("city") removes that key and returns its value; we store it under the new key.
my_dict["location"] = my_dict.pop("city")

print("After renaming city → location:")
print(my_dict)
```

### 5(ii) Add a new key `"cgpa"`

```python
my_dict["cgpa"] = STUDENT_CGPA
print("After adding cgpa:")
print(my_dict)
```

### 5(iii) Increase `"age"` by 1

```python
my_dict["age"] = my_dict["age"] + 1
print("After age += 1:")
print(my_dict)
```

### 5(iv) Delete `"branch"` with `pop()` and with `del` (on two separate copies)

```python
dict_pop = my_dict.copy()
dict_del = my_dict.copy()

returned = dict_pop.pop("branch")             # pop returns the removed value
print("After pop('branch') :", dict_pop)
print("Value returned by pop():", returned)

del dict_del["branch"]                        # del removes the key and returns nothing
print("After del ['branch'] :", dict_del)
# pop(key) returns the removed value (and can take a default); del dict[key] returns None / is a statement.
```

### 5(v) Iterate with `.items()` and print `key → value`

```python
print("Dictionary items:")
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

### 5(vi) Safe access for a missing key `"email"`

```python
if "email" in my_dict:
    print("email →", my_dict["email"])
else:
    print("Key 'email' does not exist in the dictionary. No email on file.")
```

### 5(vii) Merge with a friend's dictionary using `{**dict1, **dict2}`

```python
# friend_dict uses the same 5 *original* keys (name, roll_no, branch, age, city)
friend_dict = {
    "name": "Rohan Mehta",
    "roll_no": "1024170999",
    "branch": "Electronics and Communication Engineering",
    "age": 20,
    "city": "Pune",
}

# Merge friend's dict over mine. Rebuild my original 5-key dict so the merge matches the question.
original_mine = {
    "name": STUDENT_NAME,
    "roll_no": STUDENT_ROLL,
    "branch": STUDENT_BRANCH,
    "age": STUDENT_AGE,
    "city": STUDENT_CITY,
}

merged = {**original_mine, **friend_dict}
# When both dictionaries share a key, the second one (**friend_dict) wins — its values overwrite mine.

print("My original  :", original_mine)
print("Friend       :", friend_dict)
print("Merged       :", merged)
```

### 5(viii) Dictionary comprehension — keep only string values from the original dictionary

```python
original_dict = {
    "name": STUDENT_NAME,
    "roll_no": STUDENT_ROLL,
    "branch": STUDENT_BRANCH,
    "age": STUDENT_AGE,
    "city": STUDENT_CITY,
}

string_only = {k: v for k, v in original_dict.items() if isinstance(v, str)}

print("Original dictionary :", original_dict)
print("String-valued only  :", string_only)
```
