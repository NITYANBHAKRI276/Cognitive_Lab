```python
import pandas as pd

# Create dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)

# Save original dataset
df.to_csv("employees.csv", index=False)

# a) Shape
print("\n(a) Shape:")
print(df.shape)

# b) Summary
print("\n(b) DataFrame Summary:")
df.info()

# c) Descriptive statistics
print("\n(c) Descriptive Statistics:")
print(df.describe())

# d) First 5 and last 3 rows
print("\n(d) First 5 rows:")
print(df.head(5))

print("\nLast 3 rows:")
print(df.tail(3))

# e) Statistics
print("\n(e) Required Statistics:")
print("Average Salary:", df["Salary"].mean())
print("Total Bonus:", df["Bonus"].sum())
print("Youngest Age:", df["Age"].min())
print("Highest Rating:", df["Rating"].max())

# f) Sort by Salary descending
print("\n(f) Sorted by Salary:")
print(df.sort_values(by="Salary", ascending=False))

# g) Performance category
def categorize_performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance_Category"] = df["Rating"].apply(categorize_performance)

print("\n(g) Performance Category:")
print(df)

# h) Missing values
print("\n(h) Missing Values:")
print(df.isnull().sum())

# i) Rename Employee_ID to ID
df.rename(columns={"Employee_ID": "ID"}, inplace=True)

print("\n(i) Renamed Employee_ID to ID:")
print(df)

# j) Employee filtering
print("\n(j-i) Employees with more than 5 years experience:")
print(df[df["Years_of_Experience"] > 5])

print("\n(j-ii) Employees in IT department:")
print(df[df["Department"] == "IT"])

# k) Add Tax column
df["Tax"] = df["Salary"] * 0.10

print("\n(k) DataFrame with Tax:")
print(df)

# l) Save modified DataFrame
df.to_csv("modified_employees.csv", index=False)

print("\n(l) Modified DataFrame saved as modified_employees.csv")
```
